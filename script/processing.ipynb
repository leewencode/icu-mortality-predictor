# delete ghost patients

```python
col_count = [col for col in df.columns if col.endswith('_Count')]
```

```python
df['Total_Obs_Count'] = df[col_count].sum(axis=1)
```

# feature engineering and log transformation

### compute total missing sum and log transform count

```python
col_miss = [col for col in df.columns if col.endswith('_Miss')]
```

```python
df['Total_Missing_Sum'] = df[col_miss].sum(axis=1)
```

```python
df.shape
```

### log transformation of count variables

```python
df[col_count] = df[col_count].apply(np.log1p)
```

```python
df = df.rename(columns={col: col.replace('_Count', '_Log_Count') for col in col_count})
```

```python
df.head(20)
```

### convert _Last_Minute to _Last_Gap and compute Global_Last_Gap

```python
col_last_minute = [col for col in df.columns if col.endswith('_Last_Minute')]
```

```python
for col in col_last_minute:
    gap_name = col.replace('_Last_Minute', '_Last_Gap')
    df[gap_name] = 2880 - df[col]
```

```python
df.shape
```

```python
df = df.drop(columns=col_last_minute)
```

```python
df.shape
```

```python
col_gap = [col for col in df.columns if col.endswith('_Last_Gap')]
df['Global_Last_Gap'] = df[col_gap].min(axis=1)
```

```python
df.shape
```

### One hot encoding for ICU type and -1 for gender

```python
df['Gender'] = df['Gender'].fillna(-1)
```

```python
col_icu = [col for col in df.columns if col.startswith('ICUType_')]
df[col_icu] = df[col_icu].astype(int)
```

```python
df.shape
```

### compute expert composites: Shock_Index, GCS_MAP_Product, and Anion_Gap.

```python
df['Shock_Index'] = df['HR_Final'] / df['UniSysABP_Final']
df['GCS_MAP_Product'] = df['GCS_Final'] * df['UniMAP_Final']
df['Net_Cation_Balance'] = (df['Na_Final'] + df['K_Final']) - df['HCO3_Final']
```

# train test data split

```python
X = df.drop(columns=['In-hospital_death'])
y = df['In-hospital_death']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=SEED, stratify=y)
```

```python
X_train.reset_index(drop=True, inplace=True)
y_train.reset_index(drop=True, inplace=True)
```

```python
X_test.reset_index(drop=True, inplace=True)
y_test.reset_index(drop=True, inplace=True)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

# output tree-based model feature set

```python
X_train.to_csv(path+'/X_train_tree.csv', index=False)
X_test.to_csv(path+'/X_test_tree.csv', index=False)
y_train.to_csv(path+'/y_train.csv', index=False)
y_test.to_csv(path+'/y_test.csv', index=False)
```

# missing value imputation

### zero imputation

```python
delta_slope_cols = [c for c in X_train.columns if 'Delta' in c or 'Slope' in c]
```

```python
X_train[delta_slope_cols] = X_train[delta_slope_cols].fillna(0)
X_test[delta_slope_cols] = X_test[delta_slope_cols].fillna(0)
```

### median imputation

```python
imputer = SimpleImputer(strategy='median')
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

```python
X_train_smote = pd.DataFrame(X_train_imputed, columns=X_train.columns)
X_test_smote = pd.DataFrame(X_test_imputed, columns=X_test.columns)
```

```python
X_train_smote.isnull().sum().sum()
```

```python
X_test_smote.isnull().sum().sum()
```

```python
X_train_smote.to_csv(path+'/X_train_smote.csv', index=False)
X_test_smote.to_csv(path+'/X_test_smote.csv', index=False)
```

# standardization

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_smote)
X_test_scaled = scaler.transform(X_test_smote)
```

```python
X_train_linear = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_linear = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

# output linear-based model feature set

```python
X_train_linear.to_csv(path+'/X_train_linear.csv', index=False)
X_test_linear.to_csv(path+'/X_test_linear.csv', index=False)
```
